```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline,make_pipeline
from sklearn.preprocessing import MinMaxScaler
from sklearn.feature_selection import SelectKBest,chi2
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler
```

```python
# train_split
x_train,x_test,y_train,y_test=train_test_split(data.drop(columns=['Delyed']),data['Delyed'],test_size=0.3,random_state=42)
```

```python
x_train
```

```python
# identify which one is numerical or what is catagorical
num_cols = x_train.select_dtypes(include=['int64', 'float64']).columns
cat_cols = x_train.select_dtypes(include=['object']).columns
```

```python
num_cols
```

```python
cat_cols
```

```python
trf1 = ColumnTransformer(
    transformers=[
        ('ohe_values', OneHotEncoder(dtype=float, sparse_output=False, handle_unknown='ignore'),
         ['op_unique_carrier','origin','origin_city_name',
          'origin_state_nm','dest','dest_city_name',
          'dest_state_nm']),

        ('scaler',StandardScaler(),['year', 'month', 'day_of_month', 'day_of_week','op_carrier_fl_num', 'crs_dep_time','crs_arr_time', 'crs_elapsed_time', 'distance'])
    ],
    remainder='passthrough'
)
```

```python
# feature engineering date and time
```

```python
data['fl_date'] = pd.to_datetime(data['fl_date'])
data['flight_year'] = data['fl_date'].dt.year
data['flight_month'] = data['fl_date'].dt.month
data['flight_day'] = data['fl_date'].dt.day
data['flight_dayofweek'] = data['fl_date'].dt.dayofweek
data['is_weekend'] = data['flight_dayofweek'].isin([5, 6]).astype(int)
```

```python
x_train = x_train.drop(columns=['fl_date'])
x_test = x_test.drop(columns=['fl_date'])
```

```python
data['arr_hour'] = data['crs_arr_time'] // 100
data['arr_minute'] = data['crs_arr_time'] % 100
```

```python
# now we will use decision tree
#trf2=#DecisionTreeClassifier(max_depth=10,min_samples_split=10,min_samples_leaf=5,class_weight='balanced',random_state=42)
from sklearn.ensemble import RandomForestClassifier

trf2 = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    class_weight='balanced'
)
```

```python
# now make pipeline
pipe=make_pipeline(trf1,trf2)
```

```python
pipe
```

```python
print(x_train.columns.tolist())
```

```python
# trainning of model
pipe.fit(x_train,y_train)
```

```python
# pridiction 
y_pred = pipe.predict(x_test)
```

```python
#Accuracy cheaking
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)                        # if acc is above 70 so good 
```

```python
y_pred
```

```python
#classification report
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)
```

```python
cm                             
```

```python
from sklearn.metrics import confusion_matrix, classification_report

print(confusion_matrix(y_test, y_pred))

print(classification_report(y_test, y_pred))
```

```python
print("Training Accuracy:", pipe.score(x_train, y_train))
print("Testing Accuracy:", pipe.score(x_test, y_test))
```

```python
# save the model
import joblib

joblib.dump(pipe, "flight_delay_model.pkl")
```
